```python
import sqlite3
import pandas as pd

# Connect to database
conn = sqlite3.connect("students.db")
cursor = conn.cursor()

# Create table
cursor.execute('''CREATE TABLE IF NOT EXISTS students (id INTEGER, name TEXT)''')

# Insert data
cursor.execute("INSERT INTO students VALUES (1, 'John')")
conn.commit()
cursor.execute("INSERT INTO students VALUES (2, 'Joy')")
conn.commit()

# Read data
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)

# Update and Delete
cursor.execute("UPDATE students SET name = 'Alice' WHERE id = 1")
cursor.execute("DELETE FROM students WHERE id = 1")
conn.commit()
conn.close()
```

```python
import requests
import pandas as pd
import sqlite3

# ----------------------------
# Step 1: Fetch data from API
# ----------------------------
url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

if response.status_code == 200:
    users = response.json()

    # Convert JSON data to DataFrame
    df = pd.DataFrame(users)

    # Select required columns
    df = df[['id', 'name', 'username', 'email']]

    print("Users from API:")
    print(df)

    # ----------------------------
    # Step 2: Connect to SQLite
    # ----------------------------
    conn = sqlite3.connect("users.db")
    cursor = conn.cursor()

    # Create table
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY,
        name TEXT,
        username TEXT,
        email TEXT
    )
    """)

    # Clear previous data (optional)
    cursor.execute("DELETE FROM users")

    # Insert API data into database
    for _, row in df.iterrows():
        cursor.execute("""
        INSERT INTO users (id, name, username, email)
        VALUES (?, ?, ?, ?)
        """, (row['id'], row['name'], row['username'], row['email']))

    conn.commit()

    # ----------------------------
    # Step 3: Read data
    # ----------------------------
    print("\nData from SQLite Database:")
    db_df = pd.read_sql_query("SELECT * FROM users", conn)
    print(db_df)

    # ----------------------------
    # Step 4: Update data
    # ----------------------------
    cursor.execute("""
    UPDATE users
    SET name = 'John Smith'
    WHERE id = 1
    """)
    conn.commit()

    print("\nAfter Update:")
    updated_df = pd.read_sql_query("SELECT * FROM users", conn)
    print(updated_df)

    # ----------------------------
    # Step 5: Delete data
    # ----------------------------
    cursor.execute("DELETE FROM users WHERE id = 2")
    conn.commit()

    print("\nAfter Delete:")
    final_df = pd.read_sql_query("SELECT * FROM users", conn)
    print(final_df)

    # Close connection
    conn.close()

else:
    print("Failed to fetch data from API.")
```

```python
# Part (a): Fetch last 30 GitHub issues and move to DataFrame

import requests
import pandas as pd
import sqlite3

# GitHub API URL
url = "https://api.github.com/repos/pandas-dev/pandas/issues"

# Send GET request
response = requests.get(url)

# Convert response to JSON
issues = response.json()

# Convert JSON data to DataFrame
df_issues = pd.DataFrame(issues)

# Extract required fields
print("Last 30 GitHub Issues:")
print(df_issues[['id', 'title', 'state', 'created_at']])



# Part (b): Database Operations using SQLite

# Connect to SQLite database
conn = sqlite3.connect("employee.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employee(
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER
)
""")

# Insert records
cursor.execute("INSERT INTO employee VALUES (101, 'Rahul', 'IT', 50000)")
cursor.execute("INSERT INTO employee VALUES (102, 'Sneha', 'HR', 45000)")
cursor.execute("INSERT INTO employee VALUES (103, 'Arjun', 'Finance', 60000)")
conn.commit()

# Select data from table
print("\nEmployee Table:")
cursor.execute("SELECT * FROM employee")
rows = cursor.fetchall()

for row in rows:
    print(row)

# Move table data to DataFrame
df = pd.read_sql_query("SELECT * FROM employee", conn)

print("\nEmployee DataFrame:")
print(df)

# Update record
cursor.execute("UPDATE employee SET salary = 65000 WHERE id = 103")
conn.commit()

print("\nAfter Update:")
updated_df = pd.read_sql_query("SELECT * FROM employee", conn)
print(updated_df)

# Delete record
cursor.execute("DELETE FROM employee WHERE id = 102")
conn.commit()

print("\nAfter Delete:")
final_df = pd.read_sql_query("SELECT * FROM employee", conn)
print(final_df)

# Close connection
conn.close()
```

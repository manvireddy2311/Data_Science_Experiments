```python
# Install packages
!pip install pandas openpyxl pymupdf reportlab pillow pydub opencv-python

import pandas as pd
import fitz
from reportlab.pdfgen import canvas
from PIL import Image
from pydub import AudioSegment
import cv2

# ====================================================
# 1. CSV
# ====================================================

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("CSV Data")
print(df.head())

df.to_csv("output.csv", index=False)

print("CSV File Saved Successfully")

# ====================================================
# 2. JSON
# ====================================================

df.to_json("output.json")

df_json = pd.read_json("output.json")

print("\nJSON Data")
print(df_json.head())

print("JSON File Saved Successfully")

# ====================================================
# 3. Excel
# ====================================================

df.to_excel("output.xlsx", index=False)

df_excel = pd.read_excel("output.xlsx")

print("\nExcel Data")
print(df_excel.head())

print("Excel File Saved Successfully")

# ====================================================
# 4. PDF Write
# ====================================================

c = canvas.Canvas("output.pdf")
c.drawString(100,750,"Hello! This PDF is created using Python.")
c.save()

print("\nPDF Created Successfully")

# ====================================================
# 5. PDF Read
# ====================================================

pdf = fitz.open("sample.pdf")

print("\nPDF Content")

for page in pdf:
    print(page.get_text())

pdf.close()

# ====================================================
# 6. Image Read & Write
# ====================================================

img = Image.open("sample.jpg")

print("\nImage Size :", img.size)
print("Image Mode :", img.mode)

img.save("new_image.png")

print("Image Saved Successfully")

# ====================================================
# 7. Audio Read & Write
# ====================================================

audio = AudioSegment.from_file("sample.mp3")

print("\nAudio Duration :", len(audio), "milliseconds")

audio.export("new_audio.wav", format="wav")

print("Audio Saved Successfully")

# ====================================================
# 8. Video Read & Write
# ====================================================

cap = cv2.VideoCapture("sample.mp4")

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

fourcc = cv2.VideoWriter_fourcc(*'XVID')

out = cv2.VideoWriter("output.avi", fourcc, fps, (width, height))

while True:
    ret, frame = cap.read()

    if not ret:
        break

    out.write(frame)

cap.release()
out.release()

print("Video Saved Successfully")

print("\nAll File Operations Completed Successfully!")
```
